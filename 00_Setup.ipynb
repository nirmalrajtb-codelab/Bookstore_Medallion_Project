# Bookstore Medallion Architecture - Setup

This notebook contains common configuration, paths, schemas,
checkpoint locations, Auto Loader schema locations, table names,
and reusable functions used across the project.

Other notebooks will execute this notebook using:

%run ./00_Setup

```python
# ============================================
# PROJECT CONFIGURATION
# ============================================

PROJECT_NAME = "bookstore_medallion"

# Unity Catalog
CATALOG = "demoworkspace"

# Medallion schemas
SOURCE_SCHEMA = "source"
BRONZE_SCHEMA = "bronze"
SILVER_SCHEMA = "silver"
GOLD_SCHEMA = "gold"

print("==========================================")
print("PROJECT CONFIGURATION")
print("==========================================")

print(f"Project       : {PROJECT_NAME}")
print(f"Catalog       : {CATALOG}")
print(f"Source Schema : {SOURCE_SCHEMA}")
print(f"Bronze Schema : {BRONZE_SCHEMA}")
print(f"Silver Schema : {SILVER_SCHEMA}")
print(f"Gold Schema   : {GOLD_SCHEMA}")
```

```python
# ============================================
# CREATE SOURCE SCHEMA
# ============================================

spark.sql(f"""
CREATE SCHEMA IF NOT EXISTS {CATALOG}.{SOURCE_SCHEMA}
""")

print(f"Schema ready: {CATALOG}.{SOURCE_SCHEMA}")
```

```python
# ============================================
# CREATE MEDALLION SCHEMAS
# ============================================

spark.sql(f"""
CREATE SCHEMA IF NOT EXISTS {CATALOG}.{BRONZE_SCHEMA}
""")

spark.sql(f"""
CREATE SCHEMA IF NOT EXISTS {CATALOG}.{SILVER_SCHEMA}
""")

spark.sql(f"""
CREATE SCHEMA IF NOT EXISTS {CATALOG}.{GOLD_SCHEMA}
""")

print("Medallion schemas are ready.")
```

```python
# ============================================
# SET CURRENT CATALOG
# ============================================

spark.sql(f"USE CATALOG {CATALOG}")

print(f"Current catalog: {CATALOG}")
```

```python
# ============================================
# RAW DATA VOLUME
# ============================================

RAW_VOLUME = "bookstore_raw"

BASE_PATH = (
    f"/Volumes/{CATALOG}/{SOURCE_SCHEMA}/{RAW_VOLUME}"
)

print(f"Raw data path: {BASE_PATH}")
```

```python
# ============================================
# SOURCE DATA PATHS
# ============================================

CUSTOMERS_PATH = f"{BASE_PATH}/customers"
BOOKS_PATH = f"{BASE_PATH}/books"
ORDERS_PATH = f"{BASE_PATH}/orders"

print("Customers:", CUSTOMERS_PATH)
print("Books    :", BOOKS_PATH)
print("Orders   :", ORDERS_PATH)
```

```python
# ============================================
# CHECKPOINT PATHS
# ============================================

CHECKPOINT_BASE = f"{BASE_PATH}/checkpoints"

CUSTOMER_CHECKPOINT = (
    f"{CHECKPOINT_BASE}/customers"
)

BOOK_CHECKPOINT = (
    f"{CHECKPOINT_BASE}/books"
)

ORDER_CHECKPOINT = (
    f"{CHECKPOINT_BASE}/orders"
)

print("Customer checkpoint:", CUSTOMER_CHECKPOINT)
print("Book checkpoint    :", BOOK_CHECKPOINT)
print("Order checkpoint   :", ORDER_CHECKPOINT)
```

```python
# ============================================
# AUTO LOADER SCHEMA LOCATIONS
# ============================================

SCHEMA_BASE = f"{BASE_PATH}/schemas"

CUSTOMER_SCHEMA_PATH = (
    f"{SCHEMA_BASE}/customers"
)

BOOK_SCHEMA_PATH = (
    f"{SCHEMA_BASE}/books"
)

ORDER_SCHEMA_PATH = (
    f"{SCHEMA_BASE}/orders"
)

print("Customer schema:", CUSTOMER_SCHEMA_PATH)
print("Book schema    :", BOOK_SCHEMA_PATH)
print("Order schema   :", ORDER_SCHEMA_PATH)
```

```python
# ============================================
# BRONZE TABLES
# ============================================

BRONZE_CUSTOMERS = (
    f"{CATALOG}.{BRONZE_SCHEMA}.customers"
)

BRONZE_BOOKS = (
    f"{CATALOG}.{BRONZE_SCHEMA}.books"
)

BRONZE_ORDERS = (
    f"{CATALOG}.{BRONZE_SCHEMA}.orders"
)
```

```python
# ============================================
# SILVER TABLES
# ============================================

SILVER_CUSTOMERS = (
    f"{CATALOG}.{SILVER_SCHEMA}.customers"
)

SILVER_BOOKS = (
    f"{CATALOG}.{SILVER_SCHEMA}.books"
)

SILVER_ORDERS = (
    f"{CATALOG}.{SILVER_SCHEMA}.orders"
)
```

```python
# ============================================
# GOLD TABLES
# ============================================

GOLD_DAILY_SALES = (
    f"{CATALOG}.{GOLD_SCHEMA}.daily_sales"
)

GOLD_BOOK_SALES = (
    f"{CATALOG}.{GOLD_SCHEMA}.book_sales"
)

GOLD_CUSTOMER_SALES = (
    f"{CATALOG}.{GOLD_SCHEMA}.customer_sales"
)

GOLD_MONTHLY_REVENUE = (
    f"{CATALOG}.{GOLD_SCHEMA}.monthly_revenue"
)
```

```python
# ============================================
# DISPLAY TABLE CONFIGURATION
# ============================================

print("==========================================")
print("BRONZE TABLES")
print("==========================================")

print(BRONZE_CUSTOMERS)
print(BRONZE_BOOKS)
print(BRONZE_ORDERS)


print("\n==========================================")
print("SILVER TABLES")
print("==========================================")

print(SILVER_CUSTOMERS)
print(SILVER_BOOKS)
print(SILVER_ORDERS)


print("\n==========================================")
print("GOLD TABLES")
print("==========================================")

print(GOLD_DAILY_SALES)
print(GOLD_BOOK_SALES)
print(GOLD_CUSTOMER_SALES)
print(GOLD_MONTHLY_REVENUE)
```

```python
# ============================================
# HELPER FUNCTIONS
# ============================================

def table_exists(table_name):
    """
    Check whether a table exists.
    """
    return spark.catalog.tableExists(table_name)


def show_table_status(table_name):
    """
    Display table existence status.
    """
    if table_exists(table_name):
        print(f"✓ Table exists: {table_name}")
    else:
        print(f"✗ Table does not exist: {table_name}")
```
